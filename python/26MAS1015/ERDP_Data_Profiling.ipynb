```python
def profile_dataset(df, dataset_name):

    print("=" * 60)
    print(f"DATASET: {dataset_name}")
    print("=" * 60)

    print("\nShape:")
    print(df.shape)

    print("\nColumns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nMissing Percentage:")
    print((df.isnull().mean() * 100).round(2))

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nUnique Values:")
    print(df.nunique())

    print("\nNumerical Summary:")
    print(df.describe())
```

```python
profile_dataset(pos, "POS Transactions")
```

```python
profile_dataset(crm,"CRM Customers")
```

###Inventory - XML

```python
tree = ET.parse("/content/inventory.xml")
root = tree.getroot()
```

Converting into dictionaries

```python
inventory_data = []

for item in root:
    record = {}

    for child in item:
        record[child.tag] = child.text

    inventory_data.append(record)

inventory = pd.DataFrame(inventory_data)
```

```python
profile_dataset(inventory,"Inventory")
```

```python
summary = []

datasets = {
    "POS": pos,
    "E-Commerce": ecommerce,
    "CRM": crm,
    "Inventory": inventory,
    "Supplier": supplier
}

for name, df in datasets.items():

    summary.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing_Values": df.isnull().sum().sum(),
        "Duplicate_Rows": df.duplicated().sum()
    })

summary_df = pd.DataFrame(summary)

summary_df
```
